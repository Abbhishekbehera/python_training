```python
currency=[500,200,100]
n=int(input("Enter the number:"))
count=0
i=0
while n!=0:
    if n<currency[i]:
        i+=1
    count+=n//currency[i]
    n%=currency[i]
print(count)
```

```python
import math

n = 12
if n <= 1:
    print("Number should be greater than 1")
else:
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            print("It is not a prime number.")
        else:
            print("It is a prime number")
        break
```
